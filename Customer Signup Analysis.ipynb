```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

print("All imports successful")
```

```python
df = pd.read_csv("data/customer_signups.csv")

df.head()
```

```python
df.info()
```

```python
df.isna().sum()
```

```python
df['signup_date'] = pd.to_datetime(df['signup_date'], errors='coerce')
```

```python
df['signup_date'].dtype
```

```python
df['signup_date'] = pd.to_datetime(
    df['signup_date'],
    format='%Y-%m-%d',
    errors='coerce'
)
```

```python
df['signup_date'].dtype
```

```python
df['plan_selected'] = (
    df['plan_selected']
    .str.strip()
    .str.lower()
    .str.capitalize()
)
```

```python
df['gender'] = (
    df['gender']
    .str.strip()
    .str.lower()
    .str.capitalize()
)
```

```python
#check to see if cleaning for 'gender' and 'plan_selected' worked
df['plan_selected'].value_counts(dropna=False)
df['gender'].value_counts(dropna=False)
```

```python
#check duplicates
duplicate_count = df.duplicated(subset='customer_id').sum()
duplicate_count
```

```python
#duplicate customer IDs before
duplicate_count = df.duplicated(subset='customer_id').sum()
print(duplicate_count)
```

```python
df = df.drop_duplicates(subset='customer_id')
```

```python
#Check duplicate CustomerIDs after
duplicate_count = df.duplicated(subset='customer_id').sum()
print(duplicate_count)
```

```python
#fill missing regions with Unknown
df['region'] = df['region'].fillna("Unknown")
```

```python
df['email'] = df['email'].fillna("Unknown")
```

```python
df['age'] = pd.to_numeric(df['age'], errors='coerce')
```

```python
df['age_missing_flag'] = df['age'].isna()
```

```python
df['age_missing_flag'].sum()
```

```python
#check how many are missing an age
print(df['age_missing_flag'].sum())
```

```python
#Count of missing values per column
missing_count=df.isna().sum()
print(missing_count)
```

```python
#percentage of missing values
missing_percentage = (df.isna().mean() * 100).round(2)
missing_percentage
```

```python
duplicate_count
```

```python
print(duplicate_count)
```

```python
#table showing missing counts and percentages
data_quality_summary = pd.DataFrame({
    'Missing Count': missing_count,
    'Missing %': missing_percentage
})

data_quality_summary
```

```python
#inconsistent Categories Corrected
df['plan_selected'].value_counts(dropna=False)
```

```python
df['gender'].value_counts(dropna=False)
```

```python
signups_per_week = (
    df
    .set_index('signup_date')
    .resample('W')
    .size()
    .reset_index(name='signups')
)

signups_per_week.head()
```

```python
#sign ups by source
signups_by_source = df['source'].value_counts().reset_index()
signups_by_source.columns = ['source', 'signups']
signups_by_source.head()
```

```python
signups_by_region = df['region'].value_counts().reset_index()
signups_by_region.columns = ['region', 'signups']
```

```python
signups_by_region.head()
```

```python
signups_by_plan = df['plan_selected'].value_counts().reset_index()
signups_by_plan.columns = ['plan_selected', 'signups']
signups_by_plan.head()
```

```python
marketing_by_gender = (
    df
    .groupby(['gender', 'marketing_opt_in'])
    .size()
    .reset_index(name='count')
)

marketing_by_gender
```

```python
age_summary = df['age'].agg(
    min_age='min',
    max_age='max',
    mean_age='mean',
    median_age='median',
    null_count=lambda x: x.isna().sum()
)

age_summary
```

```python
# which source bought in the most users last month
last_month = df['signup_date'].max().to_period('M') - 1

last_month_signups = df[
    df['signup_date'].dt.to_period('M') == last_month
]

last_month_signups['source'].value_counts()
```

```python
# which region shows signs of missing or incomplete data:

df['region'].value_counts(dropna=False)
```

```python
df['age_group'] = pd.cut(
    df['age'],
    bins=[0, 24, 34, 44, 54, 100],
    labels=['<25', '25–34', '35–44', '45–54', '55+']
)

marketing_by_age = (
    df
    .groupby('age_group')['marketing_opt_in']
    .value_counts(normalize=True)
    .mul(100)
    .rename('percentage')
    .reset_index()
)

marketing_by_age
```

```python
plan_by_age = (
    df
    .groupby(['plan_selected', 'age_group'])
    .size()
    .reset_index(name='count')
)

plan_by_age
```

```python
plan_by_age.pivot(
    index='plan_selected',
    columns='age_group',
    values='count'
).fillna(0)
```
